```python
df.replace('?', np.nan, inplace = True)
```

```python
df['Country'] = df['Country'].astype(float)
df['Description'] = df['Description'].astype(float)
df['InvoiceNo'] = df['InvoiceNo'].astype(float)
df['StockCode'] = df['StockCode'].astype(float)
df['InvoiceDate'] = df['InvoiceDate'].astype(float)
```

```python
df.dtypes
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
fre_InvoiceNo = df['InvoiceNo'].dropna().mode()[0]
df['InvoiceNo'].replace(np.nan,fre_InvoiceNo,inplace=True)

fre_StockCode = df['StockCode'].dropna().mode()[0]
df['StockCode'].replace(np.nan,fre_StockCode,inplace=True)

fre_Description = df['Description'].dropna().mode()[0]
df['Description'].replace(np.nan,fre_Description,inplace=True)

fre_Quantity = df['Quantity'].dropna().mode()[0]
df['Quantity'].replace(np.nan,fre_Quantity,inplace=True)

fre_InvoiceDate= df['InvoiceDate'].dropna().mode()[0]
df['InvoiceDate'].replace(np.nan,fre_InvoiceDate,inplace=True)

fre_UnitPrice= df['UnitPrice'].dropna().mode()[0]
df['UnitPrice'].replace(np.nan,fre_UnitPrice,inplace=True)

fre_CustomerID= df['CustomerID'].dropna().mode()[0]
df['CustomerID'].replace(np.nan,fre_CustomerID,inplace=True)

fre_Country = df['Country'].dropna().mode()[0]
df['Country'].replace(np.nan,fre_Country ,inplace=True)
```
